```python
# Categorías de Yelp relacionadas con cada preferencia
relaciones = {
    "Vegetariano": ["Vegetarian"],
    "Vegano": ["Vegan"],
    "Carnes": ["Steakhouses", "Burgers", "Barbeque", "Chicken Wings"],
    "Mariscos": ["Seafood"],
    "Pescado": ["Seafood", "Sushi Bars", "Poke"]
}

# Contar las preferencias de los clientes
conteo_clientes = {
    "Vegetariano": 708,
    "Vegano": 239,
    "Carnes": 535,
    "Mariscos": 493,
    "Pescado": 272
}

# Construir la tabla comparativa
resultados = []

for preferencia, categorias_yelp in relaciones.items():

    clientes = conteo_clientes.get(preferencia, 0)

    negocios = df_restaurantes[
        df_restaurantes["categories"].apply(
            lambda x: any(c in x.split(", ") for c in categorias_yelp)
        )
    ]["id"].nunique()

    resultados.append({
        "preferencia": preferencia,
        "clientes": clientes,
        "categorias_yelp": ", ".join(categorias_yelp),
        "negocios_encontrados": negocios
    })

df_integracion = pd.DataFrame(resultados)

df_integracion
```

```python
# Relacionar categorías de Yelp con preferencias de los clientes
relacion_categorias = {
    "Vegetarian": "Vegetariano",
    "Vegan": "Vegano",
    "Steakhouses": "Carnes",
    "Burgers": "Carnes",
    "Barbeque": "Carnes",
    "Chicken Wings": "Carnes",
    "Seafood": "Mariscos, Pescado",
    "Sushi Bars": "Pescado",
    "Poke": "Pescado"
}

# Crear la columna de preferencias relacionadas
def identificar_preferencias(categorias):
    preferencias = []

    for categoria, preferencia in relacion_categorias.items():
        if categoria in categorias.split(", "):
            preferencias.extend(preferencia.split(", "))

    return ", ".join(dict.fromkeys(preferencias)) if preferencias else "Sin correspondencia"

df_restaurantes["preferencias_clientes_boston"] = (
    df_restaurantes["categories"].apply(identificar_preferencias)
)
```

```python
df_restaurantes.head(15)
```

```python
cantidad_clientes = {
    "Vegetariano": 708,
    "Vegano": 239,
    "Carnes": 535,
    "Mariscos": 493,
    "Pescado": 272
}

def contar_clientes_relacionados(preferencias):
    if preferencias == "Sin correspondencia":
        return "Sin correspondencia"

    return " | ".join(
        f"{p}: {cantidad_clientes[p]}"
        for p in preferencias.split(", ")
    )

df_restaurantes["cantidad_clientes_preferencia"] = (
    df_restaurantes["preferencias_clientes_boston"]
    .apply(contar_clientes_relacionados)
)
```
